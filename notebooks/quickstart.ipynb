## 2. Look at what the camera sees
(Exploring the simulator here is fine. Your `solution.py` must not import `fgcsim`.)

```python
import cv2
import matplotlib.pyplot as plt
from fgcsim import FieldSim

sim = FieldSim("C", 2)
obs = sim.reset()
for _ in range(60):
    obs = sim.step()
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(obs.frame, cv2.COLOR_BGR2RGB))
plt.axis("off")
print(obs.telemetry)
print(sim.camera_info)
```
